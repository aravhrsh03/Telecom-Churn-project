```python
import pandas as pd
import numpy as np
df=pd.read_csv(r'D:\projectpart1\WA_Fn-UseC_-Telco-Customer-Churn.csv')
print(df.head(5))
```

```python
print(df.isnull().sum())
print(df.dtypes)
```

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"].astype(str).str.strip(), errors="coerce")
print(df["TotalCharges"].dtypes)
print(df.isnull().sum())
```

```python
chan=df["TotalCharges"].isna() & (df["tenure"] == 0)
df.loc[chan, "TotalCharges"] = df.loc[chan,"MonthlyCharges"]
```

```python
print(df.isnull().sum())
```

```python
print("Dataset shape:", df.shape)
```

```python
print("Columns and data types:")
print(df.dtypes)
```

```python
null_counts = df.isnull().sum()
null_percent = (null_counts / len(df)) * 100
print("Null Percentage:", round(null_percent, 2), "%")
```

```python
# 4) Categorical value counts for key columns
categorical_cols = [
    "gender", "Partner", "Dependents", "PhoneService",
    "InternetService", "Contract", "PaymentMethod", "Churn"
]

for col in categorical_cols:
    if col in df.columns:
        print(f"\n{col} unique values and counts:")
        print(df[col].value_counts(dropna=False))
        print()
```

```python
# 5) Churned vs non-churned customers and churn rate
churn_counts = df["Churn"].value_counts(dropna=False)
churn_rate = (df["Churn"].eq("Yes").mean() * 100)

print("Churn counts:")
print(churn_counts)
print()
print("Churn rate (%):", round(churn_rate, 2))
print()
```
